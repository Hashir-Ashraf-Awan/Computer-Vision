```python
import cv2
import numpy as np
import matplotlib.pyplot as plt 
```

```python
img_path = "E:\\Computer Vision\\LAB\\openLAB\\test.png"
original = cv2.imread(img_path)
```

```python
cv2.imshow("Original", original)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
original_rgb = cv2.cvtColor(original, cv2.COLOR_BGR2RGB)
H, W = original.shape[:2]
cv2.imshow("Original", original_rgb)
cv2.waitKey(0)
cv2.destroyAllWindows()
gray = cv2.cvtColor(original, cv2.COLOR_BGR2GRAY)
cv2.imshow("Original", gray)
cv2.waitKey(0)

cv2.destroyAllWindows()
```

```python
#blurred = cv2.GaussianBlur(gray, (5, 5), 0)
#cv2.imshow("Blurred", blurred)
#cv2.waitKey(0)
#cv2.destroyAllWindows()
```

```python
edges = cv2.Canny(gray, threshold1=50, threshold2=150)
cv2.imshow("edges", edges)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
roi_vertices = np.array([[
    (0,H),(int(W*0.4), int(H*0.6)), (int(W*0.6), int(H*0.6)), (W,H)]], dtype=np.int32)
 
```

```python
mask = np.zeros_like(edges)
cv2.fillPoly(mask, roi_vertices, 255)
masked = cv2.bitwise_and(edges, mask)
```

```python
lines = cv2.HoughLinesP(masked, 1, np.pi/180,threshold=80,minLineLength=80,maxLineGap=60)
output = original_rgb.copy()


for line in lines:
    x1, y1, x2, y2 = line[0]
    cv2.line(output, (x1, y1), (x2, y2), (255, 15, 213), 4)
```

```python

plt.imshow(masked, cmap='gray')
plt.title("Edges + ROI")
plt.axis('off')
plt.figure(figsize=(14, 5))

plt.subplot(1, 3, 2)
plt.imshow(masked, cmap='gray')
plt.title("Edges + ROI")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(output)
plt.title("Detected Lanes")
plt.axis('off')
```
